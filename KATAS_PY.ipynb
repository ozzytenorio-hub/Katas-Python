```python
# 1. Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias 
# de cada letra en la cadena. Los espacios no deben ser considerados. 

frase_prueba = input(str("¿Que nos quieres decir?"))

def frecuencia_letras(frase):
    """frecuencia_letras 

    Esta función se encarga de mostrar la frecuencia con la que aparecen las distintas letras en una frase

    Args:
        frase (str): Es la frase que vamos a trabajar en la funcion 

    Returns:
        dicc_letras (dict): Devuelve un diccionario con las letras como variables y la frecuencia como valor
    """
    frase_limpia = frase.strip().lower() #aqui lo que hago es limpiar la frase, quitando espacios al principio y al final y normalizando todas las letras en minusculas
    dicc_letras= {} # Creamos un diccionario vacio
    for letra in frase_limpia: #Iteramos por cada uno de los elementos de la frase (letra por letra)
        if letra != ' ': #Este seria el filtro para dejar fuera los espacios
            variable = letra #convertimos cada letra en la variable, podria ser reduntante, pero me parece que asi podemos dejar mas claro, aunque se pudiese eliminar
            cantidad = frase_limpia.count(letra) # La cantidad que aparece la variable 
            dicc_letras.update({variable:cantidad})
    return dicc_letras

print(f"La frecuencia en que aparecen las letras es: {frecuencia_letras(frase_prueba)}")

# El frecuencia_letras ha sido el la primera función con la que he dado, pero me di cuenta de que si introduces numeros igual los cuenta
# entonces como el enunciado, a pesar de que no dice nada sobre los numeros, queria ir un poco mas alla así que hice la funcion frecuencia_letras2
# donde a pesar de que no lo habiamos visto en las clases investigue un poco y di con isalpha(). Tambien con el .update() investigue y di con que
# Cuenta cada letra y la va sobre escribiendo hasta que termina, mientras que el .get() la crea y si ya esta creada va añadiendo una por una. 
# En frecuencia_letras2 quité el strip, porque ya lo hace en .isalpha(), pero en frecuencia_letras si es necesario. 

def frecuencia_letras2(frase):
    """frecuencia_letras2

    Esta función se encarga de mostrar la frecuencia con la que aparecen las distintas letras en una frase

    Args:
        frase (str): Es la frase que vamos a trabajar en la funcion 

    Returns:
        dicc_letras (dict): Devuelve un diccionario con las letras como variables y la frecuencia como valor_
    """
    
    frase_limpia = frase.lower() #En este caso no hace falta hacer el strip, porque filtraremos todo lo que no sean letras con el if letra.isalpha()
    dicc_letras= {}
    for letra in frase_limpia:
        if letra.isalpha():
            dicc_letras[letra] = dicc_letras.get(letra, 0) + 1
    return dicc_letras

print(f"La frecuencia en que aparecen las letras es: {frecuencia_letras2(frase_prueba)}")


```

```python
# 2. Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map()

# Este ejercicio es bastante sencillo, con map() que funciona aplicando una funcion a cada uno de los elementos de un iterable
# Definimos nuestra lista de numeros con la libreria random que me ayuda a randomizar las cosas 
# Metemos una muestra (sample) que me permite que no se repitan dos numeros, dentro de un rango (range) y luego la cantidad de muestras 
# que quiero, en este caso 15
# defino una funcion que me multiplique un numero por dos doble_num en este caso
# luego aplicamos map generando una lista a partir de la lista original pasada por la funcion
# Al final tenemos el print de las dos listas para verlas lado a lado
# Cada vez que se corra el codigo, se genera una lista_numeros nueva, así que no hay dos resultados iguales. 

lista_numeros = random.sample(range(1, 101), 15)
print(type(lista_numeros)) # Aqui solo queria confirmar que se generara una lista

def doble_num(num): #Esta fue la primera funcion que genere para pasarle al map, era el inicio del ejercicio y lo veia como la opcion mas viable 
    """doble_num 

    Esta funcion dobla un numero

    Args:
        num (int o float): un numero 

    Returns:
        num*2 (int o float): devuelve el numero multiplicado por dos
    """
    return num*2

def doble_num_final(lista): #Mas adelante pude entender mejor como funcionaba la funcion lambda asi que regrese para hacer un second try 
    """doble_num_final 

    Esta funcion ahora dobla los numeros de una lista y devuelve una lista nueva

    Args:
        lista (list): Una lista de numeros los valores tienen que ser int o float

    Returns:
        list : devuelve una lista de numeros doblados
    """
    return list(map(lambda num: num*2, lista))

lista_numeros_dobles = list(map(doble_num, lista_numeros)) # Aqui utilizo map, para pasar la funcion original a la lista nueva 
lista_numeros_dobles_final = doble_num_final(lista_numeros) # Aqui almaceno los resultados en variables nuevas 

print(lista_numeros) # Podemos ver la lista original 
print(lista_numeros_dobles) # aqui podemos ver la funcion que funciona con el map y la funcion doble num
print(lista_numeros_dobles_final) # Podemos ver la lista doblada con la version lambda 
```

```python
# 3.Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros. La función debe
# devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.

lista_ingredientes = ["pepino", "aguacate", "cebolla", "coco", "peras", "brocoli", "cebolla roja", "cebolla japonesa"]


def filtro_lista2(palabras, busqueda):
    """filtro_lista2 
    Esta funcion toma una lista de palabras  y una palabra objetivo y consigue la palabra y devuelve una lista nueva con los elementos
    de la lista que contengan la palaabra objetivo 

    Args:
        palabras (list): Una lista de palabras type str 
        busqueda (str): la palabra que queremos buscar 

    Returns:
        list: me va a devolver una lista nueva que contenga los elementos coincidentes con la palabra
    """
    lista_limpia = [] # Creamos una lista vacia primero donde almacenaremos los parametros, antes la genere en el bucle for, y por ende el resultado
                      #no era el esperado porque la lista se generaba vacia en el bucle una y otra vez 
    for palabra in palabras: # Iteramos en la lista de palabra 
        if palabra.lower().find(busqueda.lower()) != -1: #El find devuelve el indice donde coincide, de no coincidir, devuelve -1, asi que todo resultado distinto a -1 es una coincidencia
            lista_limpia.append(palabra) #Añadimos la palabra a la lista limpia 
    return lista_limpia 

print(type(filtro_lista2(lista_ingredientes, "cebolla"))) # Confirmo que elemento es tipo lista 
print(filtro_lista2(lista_ingredientes, "cebolla")) # 3 pruebas con palabras distintas
print(filtro_lista2(lista_ingredientes, "PERAS"))
print(filtro_lista2(lista_ingredientes, "ajo"))
```

```python
# 4 Genera una función que calcule la diferencia entre los valores de dos listas. Usa la función map()

# primero Genero dos listas de numeros 
lista_numeros4 = [8, -7, 5, 9 ]
lista_numeros5 = [4.3, 5, 2.5, 8]

# Defino una funcion que reste un numero al otro 
def resta(n1, n2):
    """resta 

    Esta funcion resta dos numeros 

    Args:
        n1 (int o float): Minuendo. El primer numero, no discrimina si es positivo o negativo. 
        n2 (int o float): Sustraendo. El segundo numero que es el numero que se le restara al primero, 
                        no discrimina si es positivo o negativo asi que si es negativo hara - por - es mas

    Returns:
        int o float: Regresa la resta de los dos numeros 
    """
    return n1 - n2
# Defino una variable, tipo lista, utilizando la funcion map y le introduzco las dos listas de numero 
lista_diferencias = list(map(resta, lista_numeros4, lista_numeros5)) # Nota posterior!! Regresando a revisar me di cuenta que Este no puede ser el resultado de el ejercicio porque no es una funcion 
# Printeamos la variable lista_diferencias 
print(lista_diferencias)


# He vuelto despues a revisar y a pesar de que consigo restar dos listas, la funcion resta solo resta los dos numeros
# con la siguiente, ya esta escrito el codigo y podemos introducir cualquier lista que querramos mientras los valores sean int o float 
# Porque las operaciones aritmeticas no funcionan con str 

def diferencias(lista1, lista2):
    """diferencias 

    Resta los elementos de dos listas y devuelve una lista nueva

    Args:
        lista1 (list): lista de numeros positivos o negativos, int o float, seran el minuendo 
        lista2 (list): lista de numeros negativos, int o float, seran el sustraendo 

    Returns:
        list: regresa la lista de los dos numeros
    """
    return list(map(lambda n1, n2: n1 -n2, lista1, lista2))

print(diferencias(lista_numeros4, lista_numeros5))


# Nota posterior!! La funcion original tenia que ser: 
def diferencia_listas(lista1, lista2): #Nota posterior!! Esto es la correcion de asignar una funcion el return de esta funcion a un valor, este codigo se escribio despues de lista_diferencias
    """diferencia_listas 

Resta los elementos de dos listas y devuelve una lista nueva

    Args:
        lista1 (list): lista de numeros positivos o negativos, int o float, seran el minuendo 
        lista2 (list): lista de numeros negativos, int o float, seran el sustraendo 

    Returns:
        list: regresa la lista de los dos numeros
    """
    return list(map(resta, lista1, lista2)) #En esta funcion llamamos el map sobre la funcion que trabaja sobre los dos numeros 
print(diferencia_listas(lista_numeros4, lista_numeros5))
```

```python
# 7 Genera una función que convierta una lista de tuplas a una lista de strings. Usa la función map()
lista_tuplas = [("uno", "dos"), (3,4), (6, "siete")]
lista_mixta= ["uno", 2, "marco", 23]

def numero_a_string(lista):
    """numero_a_string 

    Convierte una lista con elementos int a string, si la lista solamente fuesen str solo habria que utilizar " ".join(lista), pero como
    tiene elementos tipo list, y .join no funciona si hay elementos distintos a str y devuelve TypeError, y ya que las tuplas pueden tener elementos mixtos
    es necesario normalizar los elementos a tipo string

    Args:
        lista (list): Admite una lista con elementos tipo int o tipo str

    Returns:
        string : Devuelve un str de los elementos de la lista
    """
    listares = []
    for elemento in lista:
        if type(elemento) == int:
            elemento = str(elemento)
        listares.append(elemento)
    nuevo_string = " ".join(listares) 
    return nuevo_string

print(numero_a_string(lista_mixta))

def funcion_tuptostr(listatuplas):
    """funcion_tuptostr 

    En una lista de tuplas, convierte todas las tuplas a strings

    Args:
        listatuplas (list): Una lista de tuplas

    Returns:
        list: devuelve una lista de strings.
    """
    lista_strings = list(map(numero_a_string, listatuplas))
    return lista_strings


print(funcion_tuptostr(lista_tuplas))

```

34. Crea la clase Arbol , define un árbol genérico con un tronco y ramas como atributos. Los métodos disponibles son:
crecer_tronco , nueva_rama , crecer_ramas , quitar_rama e info_arbol . El objetivo es implementar estos métodos para
manipular la estructura del árbol.
    
    Código a seguir:
       
        1. Inicializar un árbol con un tronco de longitud 1 y una lista vacía de ramas.
       
        2. Implementar el método crecer_tronco para aumentar la longitud del tronco en una unidad.
       
        3. Implementar el método nueva_rama para agregar una nueva rama de longitud 1 a la lista de ramas.
       
        4. Implementar el método crecer_ramas para aumentar en una unidad la longitud de todas las ramas existentes.
       
        5. Implementar el método quitar_rama para eliminar una rama en una posición específica.

        6. Implementar el método info_arbol para devolver información sobre la longitud del tronco, el número de ramas y las longitudes de las mismas.


```python
# 34

class Arbol:
    def __init__(self, nombre, tronco, ramas=None):
        self.nombre = nombre
        self.tronco = tronco
        self.ramas = ramas if ramas is not None else []
    def info_arbol(self):
        """Devuelve información básica sobre el árbol.

        Indica el nombre del árbol, la longitud del tronco y el número de
        ramas. Si no tiene ramas, lo especifica en el mensaje.

        Returns:
            str: Descripción del árbol y de sus ramas.
        """
        if self.ramas != []:
            return f"{self.nombre}, tiene un tronco de {self.tronco}m y tiene {self.ramas} ramas" 
        else:
            return f"{self.nombre}, tiene un tronco de {self.tronco}m y no tiene ramas"
    def crecer_tronco(self):
        """Aumenta en una unidad la longitud del tronco.

        Returns:
            str: Mensaje con la nueva longitud del tronco.
        """
        self.tronco += 1
        return f" El tronco de nuestro arbol ha crecido {1} metro su nueva longitud es {self.tronco} metros"
    def nueva_ramas(self):
        """Añade una nueva rama de longitud 1 al árbol.

        Returns:
            str: Mensaje indicando que se añadió la rama y el número total de ramas.
        """
        if self.ramas == []:
            self.ramas.append(1)
            return f"Nuestro {self.nombre} ya tiene su primera rama"
        else:
            self.ramas.append(1)
            return f"Nuestro {self.nombre} tiene una nueva rama, ahora tiene {len(self.ramas)} ramas"
    def crecer_ramas(self):
        """Aumenta en una unidad la longitud de cada rama del árbol.

        Returns:
            str: Mensaje con el número de ramas y sus nuevas longitudes.
        """
        self.ramas = [n + 1 for n in self.ramas] 
        return f"Ahora las {len(self.ramas)} ramas de nuestro arbol miden {self.ramas} respectivamente"
    def quitar_rama(self, posicion):
        """Elimina una rama de la posición indicada.

        Args:
            posicion (int): Índice de la rama que se eliminará. Los índices
            comienzan en 0; también se admiten índices negativos.

        Returns:
            str: Mensaje que confirma que se podó una rama.

        Raises:
            IndexError: Si la posición está fuera del rango de ramas.
        """
        self.ramas.pop(posicion)
        return f"Se ha podado una rama"
```

```python
# 36   Crea la clase UsuarioBanco ,representa a un usuario de un banco con su nombre, saldo y si tiene o no cuenta corriente. 
# Proporciona métodos para realizar operaciones como retirar dinero, transferir dinero desde otro usuario y 
# agregar dinero al saldo.

class UsuarioBanco:
    """class UsuarioBanco
    Representa a un usuario del banco y permite gestionar su saldo.

    Attributes:
        nombre (str): Nombre del usuario.
        saldo (int | float): Saldo disponible.
        cuenta_corriente (bool): Indica si el usuario tiene cuenta corriente.

    Methods:
        check_account(): Muestra si el usuario tiene cuenta bancaria.
        retirar_dinero(monto): Retira dinero del saldo; genera un ValueError si
            el monto no es válido o supera el saldo disponible.
        transferir_dinero(usuario_recibe, monto): Transfiere dinero a otro
            usuario; genera un ValueError si el monto no es válido o supera
            el saldo disponible.
        agregar_dinero(monto): Añade dinero al saldo.
    """
    def __init__(self, nombre, saldo, cuenta_corriente):
        self.nombre = nombre 
        self.saldo = saldo
        self.cuenta_corriente = cuenta_corriente
    def check_account(self):
        """check_account
        Muestra si el usuario tiene cuenta corriente en el banco.

        Returns:
            str: Mensaje con el nombre del usuario y el estado de su cuenta.
        """
        if self.cuenta_corriente is True:
            return f"{self.nombre} tiene cuenta en el banco"
        else:
            return f"{self.nombre} No tiene cuenta en el banco"
    def retirar_dinero(self, monto):
        """retirar_dinero
        Retira una cantidad de dinero del saldo del usuario.

        Args:
            monto (int | float): Cantidad que se desea retirar.

        Raises:
            ValueError: Si el monto es negativo.
            ValueError: Si el saldo es insuficiente.

        Returns:
            str: Mensaje con la cantidad retirada y el saldo restante.
        """
        if monto > 0 and self.saldo > monto: 
                self.saldo -= monto
                return f"Se ha retirado {monto}€, el saldo restante es de: {self.saldo}€"
        else: 
            if monto < 0:
                raise ValueError ("El monto no puede tener un valor negativo")
            else:
                raise ValueError ("Saldo insuficiente")
    def transferir_dinero(self, usuario_recibe, monto): 
        """transferir_dinero

        Transfiere una cantidad de dinero desde el saldo del usuario actual
        a otro usuario.

        Args:
            usuario_recibe (UsuarioBanco): El usuario que recibirá el dinero.
            monto (int | float): Cantidad a transferir.

        Raises:
            ValueError: Si el monto es negativo o si no hay saldo suficiente
            para realizar la transferencia.

        Returns:
            str: Muestra un mensaje por consola indicando el exito de la
            operación.
        """
        if monto > 0 and self.saldo > monto: 
            self.saldo -= monto
            usuario_recibe.saldo += monto
            return f"Transferencia realizada con exito"
        else:
            if monto < 0:
                raise ValueError ("El monto no puede tener un valor negativo")
            else:
                raise ValueError ("Saldo insuficiente")
    def agregar_dinero(self, monto):
        """agregar_dinero

        Añade una cantidad de dinero al saldo del usuario.

        Args:
            monto (int | float): Cantidad que se desea ingresar al saldo.

        Raises:
            ValueError: Si el monto es negativo.

        Returns:
            str: Muestra un mensaje con el saldo actualizado.
        """
        if monto > 0:
            self.saldo += monto
            return f"Aporte realizado con exito el saldo actual es de {self.saldo}€"
        else:
            print(f"El monto a ingresar no puede ser negativo")
    #Hice este ultimo metodo para revisar si las operaciones quedaban registradas, como solamente devolvia None (es decir los prints)
    #No sabia si habian quedado registradas estas operaciones 
    def revisar_saldo(self):
        """revisar_saldo 

        Hace un print con el nombre y el saldo del usuario 
        """
        return f"El saldo de {self.nombre} es de {self.saldo}€"
            
```

```python
#37 (3)

def borrar_palabra(str, palabra_borrar):
    """borrar_palabra: elimina una palabra de un texto.

    Compara las palabras sin distinguir entre mayúsculas y minúsculas y
    devuelve el texto resultante.

    Args:
        str (str): Texto del que se eliminará la palabra.
        palabra_borrar (str): Palabra que se quiere eliminar.

    Returns:
        str: Texto sin las apariciones de la palabra indicada.
    """
    str_list = str.split()
    str_list_sin_palabra = []
    for palabra in str_list:
        if palabra.lower() != palabra_borrar.lower():
            str_list_sin_palabra.append(palabra)
    nuevo_str = " ".join(str_list_sin_palabra)
    return nuevo_str
```

```python
# 39. Escribe un programa que determine qué calificación en texto tiene un alumno en base a su calificación numérica.
# Las reglas de calificación son:
# - 0 - 69 insuficiente
# - 70 - 79 bien
# - 80 - 89 muy bien
# - 90 - 100 excelente

class Alumno:
    """Representa a un alumno y permite consultar su calificación en texto.

    Attributes:
        nombre (str): Nombre del alumno.
        edad (int): Edad del alumno.
        calificacion (float): Calificación numérica, entre 0 y 100.
    """
    def __init__(self, nombre, edad, calificacion):
        self.nombre = nombre
        self.edad = edad
        self.calificacion = calificacion
    def revisar_calificacion(self):
        """revisar_calificacion
        Muestra la calificación del alumno en texto.

        Clasifica la nota en insuficiente, bien, muy bien o excelente.
        Si está fuera del rango de 0 a 100, indica que el valor es inválido.

        return (str): devuelve un str indicando la calificacion cualitativa en base a la calificación numerica
        """
        if self.calificacion >= 0 and self.calificacion <70:
            return f"Insuficiente"
        elif self.calificacion >=70 and self.calificacion <80:
            return f"Bien"
        elif self.calificacion >=80 and self.calificacion <90:
            return f"Muy bien"
        elif self.calificacion >=90 and self.calificacion <=100:
            return f"Excelente"
        else:
            return f"Revisar calificacion, valor invalido"

marcos = Alumno("Marcos", 20, 69.25)
luis = Alumno("Luis", 20, 70)
paula = Alumno("Paula", 22, 95)
maricarmen = Alumno("Maricarmen", 60, 89)

print(marcos.revisar_calificacion())
print(luis.revisar_calificacion())
print(paula.revisar_calificacion())
print(maricarmen.revisar_calificacion())
```

```python
# 40 Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo" , "circulo" o "triangulo" ) y datos (una tupla con los datos necesarios para calcular el área de la figura).

figuras_validas = ("rectangulo", "circulo", "círculo", "triangulo")

def calculo_areas(figura, valores=None): 
    """Calcula el área de un rectángulo, un triángulo o un círculo.

    Si "valores" se proporciona, debe contener las dimensiones necesarias:
    dos valores para el rectángulo o el triángulo, y un radio para el círculo.
    Si no se proporciona, solicita las dimensiones al usuario.

    Args:
        figura (str): Figura a calcular: "rectangulo", "triangulo",
            "circulo"`` o "círculo".
        valores (tuple, optional): Dimensiones necesarias para el cálculo.
            Por defecto, se solicitan mediante la entrada del usuario.

    Raises:
        TypeError: Si la figura no es válida.
        ValueError: Si "valores" no contiene la cantidad de dimensiones
            esperada para la figura.

    Returns:
        int | float: Área calculada.
    """
    valores = valores if valores is not None else None
    if figura.lower().strip() not in figuras_validas:
        raise TypeError ("figura no valida")
    elif figura == "rectangulo":
        if valores is not None:
            if len(valores) == 2:
                area = valores[0] * valores[1]
                return area
            else:
                raise ValueError("Para calcular el area del rectangulo necesitamos 2 valores")
        else: 
            base = int(input("Cual es la base del rectangulo"))
            altura = int(input("Cual es la altura del rectangulo"))
            area = base * altura 
            return area
    elif figura == "triangulo":
        if valores is not None:
            if len(valores) == 2:
                area = ((valores[0] * valores[1])/2)
                return area
            else:
                raise ValueError("Para calcular el area del triangulo necesitamos 2 valores")
        else: 
            base = int(input("Cual es la base del triangulo"))
            altura = int(input("Cual es la altura del triangulo"))
            area = ((base * altura)/2) 
            return area
    else:
        if valores is not None:
            if len(valores) == 1:
                area = (math.pi * (valores[0]**2))
                return area
            else: 
                raise ValueError("Solo necesitamos el radio del circulo")
        else:
            radio = int(input("¿Cual es el radio del circulo?"))
            area = (math.pi * (radio**2))
            return area                
```
